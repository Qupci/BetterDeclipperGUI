Restore clipped, limited and lossy-encoded peaks with [BetterDeclipper](https://github.com/Qupci/BetterDeclipperGUI) on Colab's free GPU: nothing to
install on your computer, the app opens in your browser.

1. The notebook asks for a GPU runtime already (*Runtime > Change runtime type*: **T4 GPU**). Without one,
   declipping still works on the CPU but takes minutes per song.
2. Optional: run **Connect Google Drive** to declip files from your Drive and save the results there.
3. Run **Start BetterDeclipper** (the ▶ button) and open the `https://….gradio.live` link it prints. Keep that cell
   running while you use the app; stop it to shut the app down.

```python
#@title 📁 Connect Google Drive (optional)
#@markdown Run this first to work with files in your Google Drive. In the app, type folders such as
#@markdown `/content/drive/MyDrive/Music` in the **Batch** tab's folder field (it declips every audio file
#@markdown in it) and in **Also save to folder** (results are saved there).
from google.colab import drive

drive.mount("/content/drive")
print("Your Drive is at /content/drive/MyDrive")
```

```python
#@title ▶️ Start BetterDeclipper
#@markdown Installs BetterDeclipper (about a minute) and starts it: open the **gradio.live** link that
#@markdown appears below. Keep this cell running while you use the app; stop it (■) to shut the app down.
import os
import shutil
import subprocess

gpu = ""
if shutil.which("nvidia-smi"):  # (CPU runtimes don't have it)
    gpu = subprocess.run(["nvidia-smi", "--query-gpu=name", "--format=csv,noheader"],
                         capture_output=True, text=True).stdout.strip()
print("GPU:", gpu or "none: declipping runs on the CPU and takes minutes per song "
      "(Runtime > Change runtime type > T4 GPU, then run this cell again)")
print("Installing BetterDeclipper ...")
!pip install -q --upgrade https://github.com/Qupci/BetterDeclipperGUI/archive/refs/heads/main.zip
args = "--share --no-browser"
if os.path.isdir("/content/drive/MyDrive"):
    # the folder fields read and write this machine's files, here your Drive
    args += " --allow-folders"
    print("Google Drive is connected: the folder fields reach /content/drive/MyDrive")
os.environ["PYTHONUNBUFFERED"] = "1"  # the link shows up as soon as it is there
print("Starting: open the gradio.live link below (it takes a few seconds to appear)")
!betterdeclipper-gui {args}
```

**Notes**

- Anyone who has the `gradio.live` link can use the app while it runs (with Drive connected, its folder fields
  reach your Drive), so keep the link to yourself. The link stops working when this cell is stopped.
- Colab ends idle sessions after a while and free sessions after some hours: download your results, or save
  them to Drive with *Also save to folder*.
- pip may print version conflicts with other packages that come with Colab; they don't affect the app.
